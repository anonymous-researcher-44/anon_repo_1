```python
import numpy as np
import matplotlib.pyplot as plt

def M_sqrt(M):
    U, S, _ = np.linalg.svd(M)
    
    tol = 1e-10
    rank = np.sum(S > tol)
    return U[:, :rank]
    
def orthogonal_projection_basis(Z):
    """Compute an orthonormal basis V such that VV^T equals the orthogonal projection matrix."""
    n = Z.shape[0]
    d = Z.shape[1]
    I = np.eye(n)
    P = Z @ np.linalg.solve(Z.T @ Z, Z.T)
    P_orthogonal = I - P
    
    return M_sqrt(P_orthogonal)
```

```python
np.random.seed(0)
beta = np.array([1, 2, 3])
n = 5
p = 3
xi = 10
X = np.random.normal(size=(n,p))
Z = np.random.normal(size=n)
sigma = 1.5
Y = X @ beta + Z * xi * sigma + sigma*np.random.normal(size=n)

V = orthogonal_projection_basis(X)
A = V.transpose()
mu = xi*A.dot(Z)
mu_unit_vector = mu / np.linalg.norm(mu, 2)

def get_U(Y):
    U = A.dot(Y)
    return U / np.linalg.norm(U, 2)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5), sharey=True)
for i, xi in enumerate([0, 10]):
    Us = np.array([get_U(X @ beta + Z * xi * sigma + sigma*np.random.normal(size=n)) for _ in range(200)])
    ax[i].scatter(Us[:,0], Us[:,1], s=20)
    t = np.linspace(0, 2*np.pi, 100)
    ax[i].plot(np.cos(t), np.sin(t), color='black')
    ax[i].set_ylim(-1.1,1.1)
    ax[i].set_xlim(-1.1,1.1)
    ax[i].grid(linestyle='dashed')
    ax[i].set_title(f'$\\xi = {xi}$')
    if i == 0:
        ax[i].set_ylabel('$U(Y)_2$')
    ax[i].set_xlabel('$U(Y)_1$')
    if i == 1:
        #add an arrow for mu_unit_vector
        ax[i].arrow(0,0, mu_unit_vector[0], mu_unit_vector[1], head_width=0.1, head_length=0.1, fc='red', ec='red')
plt.savefig('../figures/U_scatter.png', dpi=300)
```
