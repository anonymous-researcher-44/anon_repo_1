```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
r = 1
k = 1
n = np.arange(1,600)    # Example value, adjust as needed

alpha = 0.05  # Example value, adjust as needed

def R_function(g, n, alpha, r, k):
    nu_n = n - k
    alpha_term = alpha**(2/r)
    ratio = (alpha_term * g) / (n + g)
    power_term = ratio**(r/(nu_n + r))
    g_ratio = g / (n + g)
    
    numerator = (nu_n/r) * (1 - power_term)
    denominator = np.maximum(0, power_term - g_ratio)
    
    # Handle division by zero
    result = np.where(denominator > 1e-10, numerator / denominator, np.inf)
    return result

def R_G_function(g, n, alpha, r):

    alpha_term = alpha**(2/r)
    ratio = (g + n) / (alpha_term * g)
    result = ((g + n) / n) * np.log(ratio)
    return result


# Also create a combined plot for comparison
plt.figure(figsize=(15, 5))


# Calculate function values over the range
g=1
R_values = R_function(g, n, alpha, r, k)
R_G_values = R_G_function(g, n, alpha, r)
plt.plot(n, R_values, 'b-', linewidth=2, label=f'$R(1,n,0.05)$')
plt.plot(n, R_G_values, 'r-', linewidth=2, label=f'$R^G(1,n,0.05)$')

g=100
R_values = R_function(g, n, alpha, r, k)
R_G_values = R_G_function(g, n, alpha, r)
plt.plot(n, R_values, 'b-', linewidth=2, label=f'$R(100,n,0.05)$', linestyle='dashed')
plt.plot(n, R_G_values, 'r-', linewidth=2, label=f'$R^G(100,n,0.05)$', linestyle='dashed')

g=1000
R_values = R_function(g, n, alpha, r, k)
R_G_values = R_G_function(g, n, alpha, r)
plt.plot(n, R_values, 'b-', linewidth=2, label=f'$R(1000,n,0.05)$', linestyle='dotted')
plt.plot(n, R_G_values, 'r-', linewidth=2, label=f'$R^G(1000,n,0.05)$', linestyle='dotted')

plt.xlim(n[0],n[-1])
plt.xlabel('Sample Size (n)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
#plt.ylim(0,100)
plt.yscale('log')
plt.savefig('../figures/radius_comparison.png', dpi=300)
plt.show()
```
